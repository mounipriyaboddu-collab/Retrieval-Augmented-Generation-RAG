```python

n = int(input("Enter a number: "))
sum = 0

for i in range(1, n + 1):
    sum += i

print("Sum =", sum)
```

```python

n = int(input("Enter a number: "))

even_count = 0
odd_count = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print("Even Count =", even_count)
print("Odd Count =", odd_count)
```
